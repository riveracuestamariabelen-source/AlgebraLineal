```python
import matplotlib.pyplot as plt
from datetime import datetime

# Obtener la fecha de hoy usando la librería datetime
fecha_actual = datetime.now().strftime("%d/%m/%Y")

# ==========================================
# 1. DEFINIR LA NAVE ESPACIAL
# Diseñamos una nave en 2D usando una matriz de 2xN.
# Fila 0 son las coordenadas X, Fila 1 son las coordenadas Y.
# ==========================================
nave = [
    [0, 1, 2, 1, 3, 1, 0, -1, -3, -1, -2, -1, 0],  # Coordenadas X
    [4, 2, 1, 1, -2, -1, -3, -1, -2, 1, 1, 2, 4]   # Coordenadas Y
]

def graficar_objeto(matriz, titulo, color="cyan", ax=None):
    """Dibuja el objeto (nave) de forma cerrada en el plano cartesiano"""
    if ax is None:
        fig, ax = plt.subplots(figsize=(5, 5))

    x = matriz[0]
    y = matriz[1]

    ax.plot(x, y, marker='o', color=color, linewidth=2.5, markersize=5)
    ax.fill(x, y, color=color, alpha=0.25)

    ax.set_title(titulo, fontsize=11, color='white', fontweight='bold')
    ax.set_xlim(-15, 15)
    ax.set_ylim(-15, 15)
    ax.axhline(0, color='gray', linewidth=0.8, linestyle='--')
    ax.axvline(0, color='gray', linewidth=0.8, linestyle='--')
    ax.grid(True, linestyle=':', alpha=0.5, color='gray')

    # Estilo retro tipo videojuego de arcade (fondo oscuro)
    ax.set_facecolor('#121212')
    ax.tick_params(colors='white')
    for spine in ax.spines.values():
        spine.set_color('#444444')

# ==========================================
# 2. OPERACIONES MATRICIALES (INTERFAZ DEL JUEGO)
# ==========================================

# A) ESCALADO (Multiplicación Escalar: Aumentar o reducir tamaño del motor/escudo)
escala = 1.8
nave_escalada = [[elemento * escala for elemento in fila] for fila in nave]

# B) TRASLACIÓN (Suma de Matrices: Mover la nave a una nueva posición [X + dx, Y + dy])
dx, dy = 5, -6
matriz_traslacion = [
    [dx] * len(nave[0]),
    [dy] * len(nave[0])
]
nave_trasladada = []
for i in range(len(nave)):
    fila = []
    for j in range(len(nave[0])):
        fila.append(nave[i][j] + matriz_traslacion[i][j])
    nave_trasladada.append(fila)

# C) ROTACIÓN (Multiplicación de Matrices: Girar la nave un ángulo theta usando una matriz de rotación)
# Rotación de 90 grados en sentido antihorario:
# | cos(90)  -sin(90) |   |  0  -1 |
# | sin(90)   cos(90) | = |  1   0 |
matriz_rotacion = [
    [0, -1],
    [1,  0]
]

filas_r, cols_r = len(matriz_rotacion), len(matriz_rotacion[0])
nave_rotada = [[0] * len(nave[0]) for _ in range(filas_r)]

for i in range(filas_r):
    for j in range(len(nave[0])):
        for k in range(cols_r):
            nave_rotada[i][j] += matriz_rotacion[i][k] * nave[k][j]

# ==========================================
# 3. MOSTRAR PANEL DE EXPOSICIÓN (INTERFAZ DE SIMULACIÓN)
# ==========================================
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.patch.set_facecolor('#0a0a0a') # Fondo exterior oscuro

# Título general, Firma de Autoría y Fecha de Hoy
fig.suptitle(f"SIMULADOR DE NAVE ESPACIAL - OPERACIONES CON MATRICES\nELABORADO Por: María Rivera  |  Fecha: {fecha_actual}",
             color='white', fontsize=15, fontweight='bold', y=0.98)

# Panel 1: Estado Original
graficar_objeto(nave, "1. Posición Inicial (Nave Original)", color="cyan", ax=axs[0, 0])

# Panel 2: Zoom / Escalar
graficar_objeto(nave_escalada, f"2. Multiplicación Escalar (Zoom x{escala})", color="#FFD700", ax=axs[0, 1])

# Panel 3: Movimiento / Traslación
graficar_objeto(nave_trasladada, f"3. Suma Matricial (Mover dx={dx}, dy={dy})", color="#FF5722", ax=axs[1, 0])

# Panel 4: Dirección / Rotación
graficar_objeto(nave_rotada, "4. Mult. de Matrices (Rotación 90°)", color="#8E24AA", ax=axs[1, 1])

plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()
```
